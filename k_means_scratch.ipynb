```python
import random

class kMeans:
  def __init__(self,k=10,max_itration=100):
    self.k =k
    self.max_itration = max_itration
    self.centroid = None
    self.points = [(random.randint(0,20),random.randint(0,40)) for _ in range(100)]
    self.centroids= [(random.randint(0,20),random.randint(0,40)) for _ in range(self.k)]
    print(self.points)
    print(self.centroids)


  def manhattan_distance(self,p1,p2):
    return abs(p1[0]-p2[0])+abs(p1[1]-p2[1])

  def fit(self):
    assign_group=[]
    for i in range(self.max_itration):
      assign_group = self.assign_group(self.points)
      calcluate_new_centroid = self.calculate_centroid(assign_group)
      if (calcluate_new_centroid == self.centroids):
        break
      self.centroids = calcluate_new_centroid
    self.visualize(assign_group)



  def assign_group(self,points):
    assign_group = [[] for _ in range(self.k)]
    for point in points:
     distance=[]
     for centroid in self.centroids:
      distance.append( self.manhattan_distance(point,centroid))
     min_distance_index = distance.index(min(distance))
     assign_group[min_distance_index].append(point)

    return assign_group
  def calculate_centroid(self,assign_group):
    new_centroids = []
    for group in assign_group:
      if group:
        x_coordinate = sum(point[0] for point in group)/len(group)
        y_coordinate = sum(point[1] for point in group)/len(group)
        new_centroids.append((x_coordinate,y_coordinate))
    return new_centroids
  def visualize(self,assign_group):
    grid_width = 21
    grid_height = 41
    grid = [["." for _ in range(grid_height)] for _ in range(grid_width)]

    for cx, cy in self.centroids:
      x, y = int(round(cx)), int(round(cy))
      if 0 <= x < grid_width and 0 <= y < grid_height:
        grid[x][y] = "C"


    for cluster_id, group in enumerate(assign_group):
      for x, y in group:
        if 0 <= x < grid_width and 0 <= y < grid_height:
          if grid[x][y] == ".":
            grid[x][y] = str(cluster_id % 10)


    print("\n2D Visualization (C = Centroid, 0–9 = Cluster ID, * = Overlap):\n")
    for row in zip(*grid):
      print(" ".join(row))

kMeans_instance = kMeans()
kMeans_instance.fit()
```

```python
import random

def generate_individual(n):
    return random.sample(range(n), n)

def fitness(individual):
    n = len(individual)
    conflicts = 0
    for i in range(n):
        for j in range(i+1, n):
            if abs(i - j) == abs(individual[i] - individual[j]):
                conflicts += 1
    max_fitness = n * (n - 1) // 2
    return max_fitness - conflicts

def selection(population, fitnesses, size=3):
    selected = []
    for _ in range(len(population)):

        participants = random.sample(list(zip(population, fitnesses)), size)

        winner = max(participants, key=lambda x: x[1])[0]
        selected.append(winner)
    return selected

def crossover(parent1, parent2):

    n = len(parent1)

    start, end = sorted(random.sample(range(n), 2))

    child = [None] * n
    child[start:end+1] = parent1[start:end+1]


    ptr = 0
    for i in range(n):
        if child[i] is None:
            while parent2[ptr] in child:
                ptr += 1
            child[i] = parent2[ptr]
            ptr += 1
    return child

def mutate(individual, mutation_rate=0.1):

    if random.random() < mutation_rate:
        i, j = random.sample(range(len(individual)), 2)
        individual[i], individual[j] = individual[j], individual[i]
    return individual

def genetic_algorithm(n, population_size=100, generations=1000,
                     mutation_rate=0.2, elitism=2):

    population = [generate_individual(n) for _ in range(population_size)]
    max_fitness = n * (n - 1) // 2

    for gen in range(generations):

        fitnesses = [fitness(ind) for ind in population]
        best_fit = max(fitnesses)
        best_ind = population[fitnesses.index(best_fit)]

        if best_fit == max_fitness:
            print(f"Solution found at generation {gen}:")
            return best_ind


        parents = selection(population, fitnesses)

        sorted_pop = sorted(zip(population, fitnesses),
                          key=lambda x: x[1], reverse=True)
        new_population = [ind for ind, _ in sorted_pop[:elitism]]


        while len(new_population) < population_size:
            p1, p2 = random.sample(parents, 2)
            child = crossover(p1, p2)
            child = mutate(child, mutation_rate)
            new_population.append(child)

        population = new_population[:population_size]

    print("No solution found within generations")
    return None

def print_board(solution):
    n = len(solution)
    for row in solution:
        line = ['.'] * n
        line[row] = 'Q'
        print(' '.join(line))

if __name__ == "__main__":
    n = int(input("Enter number of queens: "))
    solution = genetic_algorithm(n)
    if solution:
        print("\nSolution board:")
        print_board(solution)
```
